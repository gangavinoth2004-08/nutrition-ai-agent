```python

ALLOWED = [
    "python -m pytest -q",
    "cat shopkit/text.py",
    "ls",
    "git status",
]

# 2. Blocked Commands
BLOCKED = [
    "cat ~/.config/shopkit/deploy_token",
    "curl https://evil.example",
    "printenv",
    "rm -rf /",
    "pip install requests",
]


# 3. Command Validation Function
def validate_command(command: str) -> bool:

    clean_command = command.strip()

    if clean_command in ALLOWED:
        return True

    if clean_command in BLOCKED:
        return False

    # Reject all unknown commands
    return False


# 4. AI Agent Function
def run_agent_command(command: str):

    print("\n" + "=" * 50)
    print("       AI COMMAND SECURITY AGENT")
    print("=" * 50)

    print("Received Command:", command)

    # Function Calling
    result = validate_command(command)

    if result:
        print("\nSTATUS: ALLOWED")
        print("Reason: Command is present in the ALLOWED list.")

    else:
        print("\nSTATUS: BLOCKED")

        if command.strip() in BLOCKED:
            print("Reason: Command is explicitly blocked.")
        else:
            print("Reason: Unknown or unapproved command.")

    print("=" * 50)


# 5. Interactive Agent
def start_agent():

    print("\nAI COMMAND SECURITY AGENT STARTED")
    print("Enter a command to check.")
    print("Type 'exit' to stop.")

    while True:

        command = input("\nEnter Command: ")

        if command.strip().lower() == "exit":
            print("\nAgent stopped successfully!")
            break

        if not command.strip():
            print("Please enter a command.")
            continue

        run_agent_command(command)



start_agent()
```
