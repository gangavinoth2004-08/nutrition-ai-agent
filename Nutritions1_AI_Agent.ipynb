### 2. Configure your Groq API Key
Before running the code below, make sure to save your Groq API key in Colab's Secrets manager on the left panel (the key icon 🔑) under the secret name **`GROQ_API_KEY`**.

```python
import os
from google.colab import userdata
from groq import Groq

# Retrieve the API key securely from Colab Secrets
try:
    groq_api_key = userdata.get('GROQ_API_KEY')
    client = Groq(api_key=groq_api_key)
    print("Groq client successfully initialized!")
except Exception as e:
    print(f"Error initializing Groq client. Make sure you set 'GROQ_API_KEY' in your Colab Secrets: {e}")
```

### 3. Build the Nutrition Agent
Let's define a function that sends a description of a meal or a nutrition question to a Groq model (like `llama-3.3-70b-versatile`) with system instructions to act as a professional nutritionist.

```python
def ask_nutrition_agent(user_prompt):
    system_instruction = ("Act as a professional")

    try:
        completion = client.chat.completions.create(
            model="openai/gpt-oss-120b", # Using a fast, high-performance Llama 3 model
            messages=[
                {"role": "system", "content": system_instruction},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.7,
            max_completion_tokens=1024
        )
        return completion.choices[0].message.content
    except Exception as e:
        return f"An error occurred: {e}"
```

### 4. Test your Nutrition Agent
You can change the test meal description below to see how the agent responds!

```python
test_meal = "hi tell me my name again"
response = ask_nutrition_agent(test_meal)
print(response)
```
